### List DS Comprehension

```python
nl = [[12, 13, 14,14], [15,16, 16, 17], [18, 19, 20]]

ls = [nl[x][y] for x in range(len(nl)) for y in range(len(nl[x]))]
print(ls)
```

```python
for i in range(len(nl)):
    for j in range(len(nl[i])):
        print(nl[i][j], end=" ")
    print()
```

```python
ls = [nl[i][j] for i in range(len(nl)) for j in range(len(nl[i]))]
print(ls)
```

### Set Comprehension

```python
print(nl)
print(ls)
st = {nl[i][j] for i in range(len(nl)) for j in range(len(nl[i]))}
print(st)

# st = {12,23,2,5,5,5,41,5} # no duplicates and unordered
# st.remove(5)
# st.pop()
```

### Dictionary Comprehension

```python
l = [1,2,3,4,5,6,7,8,9]


dc = {l[i]: l[i] for i in range(len(l))}
print(dc)

```

```python
print(nl)
dc = {nl[i][j]: nl[i][j] for i in range(len(nl)) for j in range(len(nl[i]))}
print(dc)
```
