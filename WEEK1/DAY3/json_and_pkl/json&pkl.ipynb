### JSON Serialization


```python
import json

pythondata = {
    "name":"Santhosh",
    "age": 21,
    "place": "Davangere",
    "DOB": "27-09-2004"
}

jsonData = json.dumps(pythondata)

print(jsonData, type(jsonData))
```

```python
jsonData = json.dump(pythondata, open("p1.json","x"))
```

### JSON Deserialization

Both lines read and parse JSON from p1.json, but they do it differently:


json.load() accepts a file object and reads it directly.


.read() first reads the file into a string, which json.loads() then parses.

Prefer json.load() with a context manager so the file is closed reliably:




```python
pythonData = json.loads(open("p1.json","r").read())
pythonData
```

```python
pythonData = json.load(open("p1.json","r"))
pythonData
```

### PKL Serialization

```python
import pickle as pkl

pythondata = "This is a dummy paragraph for demonstration purposes."

pickle_data = pkl.dumps(pythondata)
pickle_data
```

```python
import pickle as pkl

pythondata = "This is a dummy paragraph for demonstration purposes."

pickle_data = pkl.dump(pythondata , open("pickle_data.pkl", "wb"))
```

```python
import pickle as pkl
fl = ["This is a dummy paragraph for demonstration purposes."]
pickle_data = pkl.dump(fl, open("pickle_data1.pkl", "wb"))
pickle_data
```

### PKL deserialization

```python
pythondata = pkl.load(open("pickle_data1.pkl", "rb"))
pythondata
```

```python
pythondata = pkl.loads(open("pickle_data1.pkl", "rb").read())
pythondata
```
