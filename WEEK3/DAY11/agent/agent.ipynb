```python
from langchain_google_genai import ChatGoogleGenerativeAI,GoogleGenerativeAIEmbeddings
from langchain_core.tools import tool
from langchain.agents import create_agent


@tool #tool(get_word_count)
def get_word_count(text: str) -> str:
    """Count the number of words in the given text."""
    return f"{len(text.split())} words"
@tool
def reverse_text(text: str) -> str:
    """Reverse the given text."""
    return text[::-1]
tools = [reverse_text,get_word_count]
# tools = [get_word_count, reverse_text]
# ── LLM ─────────────────────
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash",api_key="key")
# ── Create Agent ─────────────────────
agent = create_agent(llm, tools)
# ── Run Agent ─────────────────────
response = agent.invoke({
    "messages": [
        ("user", "How many words are in 'Hello world this is LangChain'?")
    ]
})
print(response)
print(response["messages"][-1].content)
```

### 1. Defining Tools
**What is a "Tool"?** 
In the world of AI Agents, a **Tool** is simply a Python function that the LLM is allowed to execute. While an LLM can talk, it cannot "do" things (like check a database, calculate a complex math problem, or search the web). A tool gives the LLM "hands" to interact with the real world.

**The Implementation**:
We create standard Python functions (`get_word_count` and `reverse_text`) and mark them with the `@tool` decorator. 

**Why use tools?**
LLMs often struggle with precise logic (like counting exactly how many words are in a sentence). By providing a `get_word_count` tool, the AI doesn't "guess" the count—it calls the Python function to get the **exact** mathematical answer.

**The Secret Sauce: The Docstring**:
The text inside `"""..."""` is not just a comment; it is an instruction for the AI. The LLM reads these docstrings to decide: *"Does this user want to count words? Yes, I have a tool with a description that says it counts words. I will call that!"*

# 🤖 AI Agents with LangChain
This notebook demonstrates the implementation of an **AI Agent** that can use external tools to perform tasks that a standard LLM cannot do on its own (like precise counting or text manipulation).

## 🎯 What is an AI Agent?
A standard LLM predicts the next token based on patterns. An **Agent** is an LLM wrapped in a loop that can:
1. **Reason**: Decide which tool is needed to solve a problem.
2. **Act**: Execute a specific tool (function).
3. **Observe**: Read the tool's output and decide if the task is finished or if another tool is needed.

## 🛠️ The Workflow in this File
- **Tools**: Custom Python functions decorated with `@tool` that the agent can "call."
- **Brain**: Google Gemini (the LLM) which decides which tool to use.
- **Executor**: `create_agent` which connects the brain to the tools.
