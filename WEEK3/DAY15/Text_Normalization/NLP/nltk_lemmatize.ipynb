# Lemmatization
**What is it?**
Lemmatization is a text normalization technique that reduces a word to its base or dictionary form, known as the **lemma**. Unlike stemming (which simply chops off the ends of words), lemmatization uses vocabulary and morphological analysis to ensure the resulting word is a valid word in the language.

**Why use it?**
1. **Accuracy:** It understands the context and meaning of the word (e.g., "better" $\rightarrow$ "good").
2. **Normalization:** It groups together different inflections of the same word (e.g., "mice" $\rightarrow$ "mouse", "children" $\rightarrow$ "child"), which reduces the dimensionality of the data and improves the performance of NLP models.
3. **Meaning Preservation:** Because it uses a dictionary, it avoids creating non-words (which stemming often does).

```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
tokens = ['running', 'cars', 'leaves', 'geese', 
          'better', 'mice', 'boxes', 'children', 
          'wolves', 'studies']
# lemmatizer.lemmatize(tokens) --> type error

for word in tokens:
    print(word, lemmatizer.lemmatize(word))
```
