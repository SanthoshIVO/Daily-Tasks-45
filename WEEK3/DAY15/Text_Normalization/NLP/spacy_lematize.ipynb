# NLP Fundamentals: Spacy vs NLTK
This notebook focuses on using **spaCy** for text normalization. To understand why we use it, let's look at the core concepts.

### 1. What is Tokenization?
In simple words: **It is the process of breaking a big piece of text into smaller pieces (tokens).**
*   **Word Tokenization:** Splitting a sentence into words. (`"I love AI"` $\rightarrow$ `["I", "love", "AI"]`)
*   **Sentence Tokenization:** Splitting a paragraph into sentences.

### 2. NLTK vs. spaCy
Both are popular libraries for Natural Language Processing (NLP), but they work differently:

| Feature | **NLTK** (Natural Language Toolkit) | **spaCy** |
| :--- | :--- | :--- |
| **Approach** | A collection of many different algorithms for the same task. | Provides one "best-in-class" optimized pipeline. |
| **Speed** | Slower (more academic/research oriented). | Very fast (designed for industrial production). |
| **Usage** | You have to manually call `tokenize` $\rightarrow$ `pos_tag` $\rightarrow$ `lemmatize`. | You just call `nlp(text)` and it does everything automatically. |
| **Lemmatization** | Needs a dictionary (WordNet) and POS tags to be accurate. | Uses a built-in pipeline and pre-trained models for high accuracy. |

### 3. Recap of Previous Concepts
As we've seen in the previous notebooks:
*   **Stemming:** "Chopping" off the end of words (fast but sometimes creates non-words).
*   **Lemmatization:** Finding the "dictionary root" of a word (accurate, understands context).
*   **Stopwords:** Removing common words (like "the", "is") that don't add much meaning.
*   **Chunking:** Grouping words into meaningful phrases (like "The big brown dog") instead of looking at single words.

```python
import spacy
nlp = spacy.load('en_core_web_sm')
tokens = nlp('Apply the pipeline to some text. The text can span multiple sentences, '
                'and can contain arbitrary whitespace. Alignment into the original ' \
                'string is preserved.')
print(type(tokens))

```

```python
lematize_tokens = [token.lemma_ for token in tokens if not token.is_stop]
for token in tokens:
    print(type(token))
print(tokens.lemma_,tokens.is_stop)
print(" ".join(lematize_tokens))
```
