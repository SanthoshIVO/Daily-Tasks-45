# Stopwords Removal
**What is it?**
Stopwords are common words in a language (like "the", "is", "in", "at", "which", "on") that appear frequently but usually carry very little meaningful information for NLP tasks. Stopword removal is the process of filtering these words out of the text.

**Why use it?**
1. **Reduce Noise:** By removing words that don't add specific meaning, the model can focus on the important "keywords" that define the topic of the text.
2. **Efficiency:** It reduces the number of tokens the model needs to process, which saves memory and speeds up computation.
3. **Improve Accuracy:** In tasks like document classification or search, removing stopwords prevents the model from incorrectly associating documents based only on common words.

```python
# Import necessary libraries and download the stopwords dataset
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords

nltk.download('stopwords')
nltk.download('punkt') # Needed for tokenization
stop_words = stopwords.words('english')
```

```python
# Read text data from a local file
with open('data.txt', 'r') as file:
    data = file.read()
```

```python
# Convert the raw text into a list of individual words (Tokenization)
tokens = word_tokenize(data)
```

```python
# Remove stopwords from the tokenized list
updated_tokens = []
for words in tokens:
    if words not in stop_words:
        updated_tokens.append(words)
print(updated_tokens)
```

```python
# An alternative (though less precise) way to remove stopwords by replacing them with spaces in the original string
for word in stop_words:
    if word in data:
        data = data.replace(word, ' ')
print(data)
```

```python
# Demonstrate Sentence Tokenization and filtering stopwords from sentences
sent_tokens = sent_tokenize(data)
print(sent_tokens)
updated_tokens = []
for words in sent_tokens:
    if words not in stop_words:
        updated_tokens.append(words)
print(updated_tokens)
```
