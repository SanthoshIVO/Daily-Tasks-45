```python
'''Fixed size Chunking'''


with open('data.txt') as file:
    data = file.read()
# print(data)
chunks=[]
fixed_size = 20
for index in range(0, len(data), fixed_size):
    chunk = data[index:index+fixed_size]
    print(index, end= ' ')
    # print(chunk, len(chunk))
    chunks.append(chunk)
print(chunks)
```

```python
''' Fixed size + overlapping '''

with open('data.txt') as file:
    data = file.read()
chunks=[]
fixed_size = 20
overlap_size = 4
for index in range(0, len(data), overlap_size):
    chunk = data[index:index+fixed_size]
    print(index, end= ' ')
    # print(chunk, len(chunk))

    chunks.append(chunk)

print(chunks)

```
