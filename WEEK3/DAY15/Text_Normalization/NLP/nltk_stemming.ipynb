# Stemming
**What is it?**
Stemming is a text normalization technique that reduces a word to its stem or root form by removing suffixes. It uses a set of heuristic rules (like "remove -ing", "remove -ed") to chop off the ends of words. Unlike lemmatization, it does not use a dictionary and does not consider the word's part of speech.

**Why use it?**
1. **Speed:** It is computationally very fast because it follows simple rules.
2. **Simplicity:** It helps in grouping similar words (e.g., "running", "runs", "runner" $\rightarrow$ "run") to reduce the total number of unique words in a dataset.
3. **Search Efficiency:** Useful in search engines where you want to find all variations of a root word without needing a full linguistic analysis.

```python
# Import PorterStemmer from nltk
# PorterStemmer is one of the oldest and most common stemming algorithms
from nltk.stem import PorterStemmer
```

```python
# Initialize the PorterStemmer object
stemmer = PorterStemmer()
```

stemmer works in two ways
1) it converts data to lower case
2) based on stemmer algorith in applies stemming logic

```python
# Define a list of words to test the stemmer
words = [
    "Running", "jumps", "easily", "flying", "happier",
    "Organization", "computational", "studies", "relational",
    "running", "better", "leaves", "cities", "studying",
    "connections", "trusted", "friendly", "learning",
    "management", "national"
]

# Apply the stemmer to each word and print the result
for word in words:
    print(f'{word}---->{stemmer.stem(word)}')
```

in above o/p we can observe the stemming is done, but inproper for some words
in that case we use **lemmatization**

```python
# Import SnowballStemmer, which is more advanced than PorterStemmer
# It supports multiple languages and provides better stemming results
from nltk.stem import SnowballStemmer
snowball_stemmer = SnowballStemmer('english', ignore_stopwords=True)
```

```python
# Define words for SnowballStemmer and apply stemming
snowball_words = [
    "running", "jumps", "easily", "flying", "happier",
    "organization", "computational", "studies", "relational",
    "better", "leaves", "cities", "studying", "connections",
    "trusted", "friendly", "learning", "management",
    "national", "analysis"
]

for word in snowball_words:
    print(f'{word}--->{snowball_stemmer.stem(word)}')
print(word)
```

```python
# Manually overriding the stopwords list in SnowballStemmer
snowball_stemmer.stopwords = ['a','the']
print(snowball_stemmer.stopwords)
```

```python
# Importing nltk stopwords and applying them to SnowballStemmer
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer
stop_words = stopwords.words("english")

# Initialize SnowballStemmer and set its stopwords
snowball_stemmer = SnowballStemmer('english')
snowball_stemmer.stopwords = stopwords

# Stem the words
for word in snowball_words:
    print(word,snowball_stemmer.stem(word))
```
