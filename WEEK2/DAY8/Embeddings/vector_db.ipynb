```python
from sentence_transformers import SentenceTransformer
from langchain_text_splitters import RecursiveCharacterTextSplitter
import spacy
```

```python
# %pip install sentence_transformers
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
data = open('data.txt').read()
data
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 20
)
```

```python
chunks = splitter.split_text(data) #list of strings(chunks it returns)
chunks
```

```python
embedding_model = SentenceTransformer(
    model_name_or_path='sentence-transformers/all-miniLM-L6-V2'
)
```

```python
chunk_embeddings = embedding_model.encode(chunks, convert_to_numpy=True).astype('float32')
chunk_embeddings.shape
```

```python
chunk_embeddings
# chunks
# dimension = chunk_embeddings.shape
```

Search Method(R)

```python
Query = 'what is Machine Learning'
query_embeddings = embedding_model.encode(Query).astype('float32')
query_embeddings = query_embeddings.reshape(1, -1)
faiss.normalize_L2(query_embeddings)
```

```python
index = faiss.IndexFlatIP(chunk_embeddings.shape[1])
index.add(chunk_embeddings)
# index = faiss.IndexFlat()
```

```python
'''
1st value = distance
2nd value = index of the chunk
'''
index.search(query_embeddings,k=4)
```

```python
# write a user derfined function name rag_query(query), this function accept query and returns distance and index
def rag_query(query, k=4):
    query_embeddings = embedding_model.encode(query).astype('float32')
    query_embeddings = query_embeddings.reshape(1, -1)
    faiss.normalize_L2(query_embeddings)
    distances, indices = index.search(query_embeddings, k)
    return distances, indices
rag_query(Query)
```
