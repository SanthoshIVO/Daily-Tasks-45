# ChromaDB with LangChain and Gemini embeddings

This notebook creates a ChromaDB vector store using a LangChain configuration instead of calling ChromaDB directly.

```python
import os
from dotenv import load_dotenv

load_dotenv()
api_key = os.getenv("GEMINI_API_KEY") or os.getenv("GOOGLE_API_KEY")
if not api_key:
    raise ValueError("Set GEMINI_API_KEY or GOOGLE_API_KEY in your environment or .env file.")
```

```python
from langchain_chroma import Chroma
from langchain_core.documents import Document
from langchain_google_genai import GoogleGenerativeAIEmbeddings

texts = [
    "How do I write a Python function?",
    "How can I create a class in Java?",
    "Which smartphone has the best camera?",
    "What is the best Android phone?",
    "Turn off the lights.",
    "Switch the lights off.",
    "Apple released a new iPhone.",
    "Apple is my favorite fruit."
]

documents = [Document(page_content=text) for text in texts]

embedding_model = GoogleGenerativeAIEmbeddings(
    model="models/text-embedding-004",
    google_api_key=api_key,
    task_type="retrieval_document"
)

vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embedding_model,
    collection_name="santhosh-langchain-collection",
    persist_directory="./chromadb_langchain_db"
)

print(f"Stored {len(documents)} documents in ChromaDB.")
```

```python
query = "Which phone has the best camera?"
results = vectorstore.similarity_search(query, k=3)

print("Query:", query)
print("---")
for result in results:
    print(result.page_content)
    print("---")
```
